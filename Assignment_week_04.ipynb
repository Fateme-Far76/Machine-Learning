```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read the file
with open('housing-data.txt', 'r') as file:
    content = file.readlines()

# Extract column names from the first row
column_names = content[0].strip().split(',')

# Create a list of dictionaries for each row (excluding the first row)
data = []
for line in content[1:]:

    # Split the line into columns
    columns = line.strip().split(',')

    # Create a dictionary for the row
    row = {}
    for i, col_name in enumerate(column_names): 

        # Convert column value to numeric type
        row[col_name] = pd.to_numeric(columns[i])
        
    # Add the row to the data list
    data.append(row)

# Create the DataFrame
df = pd.DataFrame(data)
df
```

```python
def compute_cost(X, y, theta):
    m = X.shape[0]  
    h = np.dot(X, theta)
    cost = np.sum((h - y) ** 2) / (2 * m)
    return cost
```

```python
def gradient_descent(X, y, theta, alpha, num_iters):
    m = X.shape[0] 
    ones_column = np.ones((m, 1))
    X = np.concatenate((ones_column, X), axis=1)
    
    costs = []  # List to store the costs
    
    for _ in range(num_iters):
        h = np.dot(X, theta)
        error = h - y
        gradient = np.dot(X.T, error) / m
        theta = theta - alpha * gradient
        
        # Calculate the cost and append it to the list
        cost = compute_cost(X, y, theta)
        costs.append(cost)
    
    return theta, costs
```

```python
# Initialize some parameters 
X = df['size'].values.reshape(-1, 1)  
y = df['price'].values.reshape(-1, 1)  
 
theta = np.zeros((2, 1))
alpha = 0.1
num_iters = 1000

# Run gradient descent
updated_theta, cost_history = gradient_descent(X, y, theta, alpha, num_iters)

# Plot the cost history
plt.plot(range(num_iters), cost_history)
plt.xlabel('Iteration')
plt.ylabel('Cost')
plt.title('Cost History')
plt.show()
```
